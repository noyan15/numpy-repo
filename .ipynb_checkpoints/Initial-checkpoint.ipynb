```python
import math
arr = np.array([1,4,3,2,6,7,1,2,3,1,2,3,3,2,3,2,2,4,6,7,8,4,3,2,4])

lnth = arr.size
print(lnth)
sqr_lnth = np.sqrt(lnth)

rows = math.ceil(sqr_lnth)
colums = math.floor(sqr_lnth)

r_ar = arr.reshape(rows,colums)

print(r_ar)
```
