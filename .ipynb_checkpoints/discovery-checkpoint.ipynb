```python
import numpy as np
import math
arr = np.array([1,4,3,2,6,7,1])

lnth = arr.size
sqr_lnth = np.sqrt(lnth)

print(lnth)

def rtrn_valid_sizes(recall,num,ln):
    r = math.ceil(ln/num)
    c = colums = math.floor(ln/r)
    if(r*c==ln):
        return ({"colums":c,"rows":r})
    else:
        return (recall(recall,num+1,ln))

if(math.ceil(sqr_lnth) * math.floor(sqr_lnth) == lnth):
    rows = math.ceil(sqr_lnth)
    colums = math.floor(sqr_lnth)
elif(math.ceil(sqr_lnth+1) * math.floor(sqr_lnth) == lnth):
    rows = math.ceil(sqr_lnth+1)
    colums = math.floor(sqr_lnth)
else:
    rtrn_values = rtrn_valid_sizes(rtrn_valid_sizes,2,lnth)
    rows = rtrn_values["rows"]
    colums = rtrn_values["colums"]


r_ar = arr.reshape(rows,colums)
print(r_ar)

r_ar = arr.reshape(-1)
print(r_ar)
```
